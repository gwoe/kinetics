```python
import pubchempy as pcp
import pandas as pd

molecules = ["Hydrogen", "Helium", "Nitrogen", "Oxygen", "Argon"]

data = []

for mol in molecules:
    compound = pcp.get_compounds(mol, 'name')[0]
    data.append({
        "Molecule": mol,
        "Molecular Mass": compound.molecular_weight
    })

df = pd.DataFrame(data)
df
```

```python
masses = {
    "H₂": 2.016e-3 / 6.02214076e23,
    "He": 4.0026e-3 / 6.02214076e23,
    "N₂": 28.014e-3 / 6.02214076e23,
    "O₂": 31.998e-3 / 6.02214076e23,
    "Ar": 39.948e-3 / 6.02214076e23
}
```

```python
def maxwell_pdf(v, T, m):
    sigma = np.sqrt(Boltzmann * T / m)
    return np.sqrt(2/np.pi) * (v**2 / sigma**3) * np.exp(-v**2/(2*sigma**2))
```

```python
def mean_velocity(T, m):
    return np.sqrt(8 * Boltzmann * T / (np.pi * m))
```

```python
def plot_distribution(T, gas):

    m = masses[gas]

    v = np.linspace(0, 6500, 2000)
    pdf = maxwell_pdf(v, T, m)

    v_mean = mean_velocity(T, m)

    plt.figure(figsize=(8,5))
    plt.plot(v, pdf)

    plt.fill_between(v, pdf, alpha=0.3)

    plt.scatter([v_mean],[0], color="blue", s=80)

    plt.xlabel("v [m/s]")
    plt.ylabel("density function")
    plt.title("Velocity distribution")

    plt.text(3500, max(pdf)*0.5,
             f"temperature: {T:.0f} K = {T-273:.1f} °C",
             fontsize=12)

    plt.xlim(0,6500)
    plt.show()
```

# Interactive plot

```python
T_slider = widgets.FloatSlider(
    value=293,
    min=150,
    max=1200,
    step=10,
    description="Temperature (K)"
)

gas_menu = widgets.Dropdown(
    options=list(masses.keys()),
    value="N₂",
    description="Gas"
)

widgets.interactive(plot_distribution, T=T_slider, gas=gas_menu)
```
