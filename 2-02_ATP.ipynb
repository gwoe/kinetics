```python
from pint import UnitRegistry
u = UnitRegistry()
Q = u.Quantity
```

```python
n_A = Q(6.02214129e23, 1/u.mole)     # Avogadro constantureg.formatter.default_format = "P"
k_B = Q(1.380649e-23, u.joule/u.kelvin)
```

## Constants and values

<p>Molecular weights of ADP and ATP</p>

```python
mATP = Q(507.18, u.gram/u.mole)
mADP = Q(427.201, u.gram/u.mole)
```

<p>The following concentrations of ATP + ADP in erythrocytes have been reported: </p>

```python
cATP_erythrocytes = Q(2.5, u.millimole/u.liter)
cATP_erythrocytes
```

```python
cATP_muscle = Q(5.0, u.millimole/u.liter)
cATP_muscle
```

<p>We assume that the person weighs 72 kg, and that the body density is app. 1.2 g/ml, and calculate the body volume: </p>

```python
body_density = Q(1.2, u.gram/u.milliliter)
body_mass = Q(72.0, u.kilogram)
```

```python
body_vol = (body_mass/body_density).to(u.liter)
body_vol
```

We average the masses of ATP and ADP, assuming 80% ATP and 20% ADP

```python
m_mean = (0.8 * mATP + 0.2 * mADP)/1.0
f"{m_mean:.2f~P}"
```

```python
ATP_mass_ery = (cATP_erythrocytes * m_mean * body_vol)
f"{ATP_mass_ery.to(u.gram):.3f~P}"
```

```python
ATP_mass_muscle = (cATP_muscle * m_mean * body_vol)
f"{ATP_mass_muscle.to(u.gram):.3f~P}"
```

<p>The reported free enthalpies ∆G for ATP synthesis and hydrolysis are: </p>

```python
deltaG_synth = Q(50.0, u.kilojoule/u.mole)
deltaG_hydrol = Q(30.5, u.kilojoule/u.mole)
```

<p>...and hence </p>

```python
(deltaG_hydrol / n_A).to(u.joule)
```

```python
f"{(deltaG_hydrol / n_A).to(u.joule):~P}"
```

### Unit conversions

```python
((deltaG_hydrol * Q(1, u.mole)).to(u.newton*u.meter) / (6.022e23))
```

```python
((deltaG_hydrol * Q(1, u.mole)).to(u.joule) / (6.022e23))
```

## Daily turnover

<p>A 'standard' woman needs app. 6,100 kJ/day, a man ~6,800 kJ/day, which converts to daily grams of ATP turned over:</p>

```python
daily_energy = Q(6800., u.kilojoule/u.day)
ATP_turnover = (mATP * daily_energy) / deltaG_synth
f"{(ATP_turnover.to(u.kilogram/u.day)):.2f~P}"
```

<h3>Result:</h3>
<p>Based on ref. (1), the <span style="color:darkblue">daily nutrition energy</span> of 6,100 (♀︎) or 6,800 (♂︎) kJ/day requires a daily turnover of <span style="color:darkblue">60 kg of ATP</span>.</p>

(1) Roza, A. M. & Shizgal, H. M. The Harris Benedict equation reevaluated: resting energy requirements and the body cell mass. Am J Clin Nutr 40, 168-182 (1984).
