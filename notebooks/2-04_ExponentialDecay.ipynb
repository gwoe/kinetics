```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
```

```python
def plot_decay(isize, grid, thn, color, fs, k):

    t = np.linspace(0, 10, 600)

    exp_decay = np.exp(-k*t)
    linear_approx = 1 - k*t

    plt.figure(figsize=(isize/100, isize/150))

    plt.plot(t, exp_decay, color=color, linewidth=thn*1000)
    plt.plot(t, linear_approx, color="gray", linewidth=1)
    plt.plot(t, np.full_like(t,0.5), "--", color="gray")
    plt.plot(t, np.full_like(t,1/np.e), "--", color="gray")

    plt.xlim(0,5)
    plt.ylim(0,1)

    if grid:
        plt.grid(True)

    plt.xlabel("time / s", fontsize=fs)
    plt.ylabel("concentration", fontsize=fs)

    plt.title("Single-exponential decay", fontsize=fs+2)

    lifetime = 1/k
    half_life = np.log(2)/k

    plt.scatter([lifetime],[0.01], s=80)
    plt.scatter([half_life],[0.5], s=80)
    plt.scatter([lifetime],[1/np.e], s=80)

    plt.text(0.55+lifetime,0.075,
             f"Decay constant: {k:.2f} s⁻¹\nLifetime: {lifetime:.2f} s",
             fontsize=fs)

    plt.text(0.75+half_life,0.525,
             f"Half-life time: {half_life:.2f} s",
             fontsize=fs)

    plt.show()
```

```python
widgets.interactive(
    plot_decay,
    isize=widgets.Dropdown(options=[450,600,720,950,1080,1920], value=720),
    grid=widgets.Dropdown(options=[("None",False),("Automatic",True)], value=True),
    thn=widgets.Dropdown(options=[0.002,0.0035,0.005,0.0067,0.008,0.01], value=0.0035),
    color=widgets.Dropdown(options=["black","blue","lightblue","gray","lightgray","green","lightgreen"], value="blue"),
    fs=widgets.Dropdown(options=[6,9,12,14,18,24], value=12),
    k=widgets.FloatSlider(value=1,min=0.25,max=5,step=0.1,description="k")
)
```

<h3>Figure legend</h3><p><span style="color:blue">Blue</span>: exponential curve $f(x)=e^{-k \cdot t}$, upper <span style="color:darkgray">dotted line</span>: 50% line, lower <span style="color:darkgray">dotted line</span>: $\frac{1}{e} \approx 0.3679$</p>
