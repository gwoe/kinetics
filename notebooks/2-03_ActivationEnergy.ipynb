```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
```

<p>Give a graphical impression of the energy landscape of a chemical reaction</p>

```python
def reaction_profile(x, deltaG, Eactivation, width, dx):
    return np.exp(-deltaG * x) + Eactivation * np.exp(-width * (x - dx)**2)
```

```python
def plot_profile(color, grid, ftsize, thn, isize,
                 Eactivation, dx, width, deltaG, yRange):

    x = np.linspace(0, 2.5, 600)
    y = reaction_profile(x, deltaG, Eactivation, width, dx)

    plt.figure(figsize=(isize/100, isize/160))

    plt.plot(x, y, color=color, linewidth=thn*1000)

    if grid:
        plt.grid(True)

    plt.xlabel("reaction coordinate", fontsize=ftsize)
    plt.ylabel("ΔG", fontsize=ftsize)

    plt.ylim(0, yRange)

    plt.tick_params(labelsize=ftsize)

    plt.show()
```

```python
color_widget = widgets.Dropdown(
    options=["blue","lightblue","green","red","orange","yellow","gray","white"],
    value="blue",
    description="Color"
)

grid_widget = widgets.Dropdown(
    options=[("None", False), ("Automatic", True)],
    value=False,
    description="Grid"
)

ft_widget = widgets.Dropdown(
    options=[12,14,18,21,24,28,36],
    value=18,
    description="Font size"
)

th_widget = widgets.Dropdown(
    options=[0.0025,0.004,0.0067,0.0075,0.01],
    value=0.004,
    description="Thickness"
)

size_widget = widgets.Dropdown(
    options=[450,650,720,900,1050,1280,1920],
    value=720,
    description="Image size"
)

widgets.interactive(
    plot_profile,
    color=color_widget,
    grid=grid_widget,
    ftsize=ft_widget,
    thn=th_widget,
    isize=size_widget,
    Eactivation=widgets.FloatSlider(value=1,min=0,max=5,step=0.1,description="Ea"),
    dx=widgets.FloatSlider(value=1,min=0.5,max=2,step=0.05,description="δ"),
    width=widgets.FloatSlider(value=5,min=1,max=10,step=0.5,description="width"),
    deltaG=widgets.FloatSlider(value=1,min=0,max=5,step=0.1,description="ΔGrxn"),
    yRange=widgets.FloatSlider(value=1.5,min=0.5,max=5,step=0.1,description="y-range")
)
```
