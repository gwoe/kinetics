<h1>Competing Substrates</h1>
<h3>Substrate and substrate analog</h3>
<p>G. Woehlke 2017ff (converted from Wolfram Mathematica, aided by A.I.)</p>

<p>We assume the following kinetic scheme (Segel, p. 127):</p>
<img src="2-13_CompetingSubstrates.png" width="466"> 
<p><ul>
    <li>E	enzyme
    <li>Sa	substrate A
<li>Sb	substrate B
<li>Pa, Pb	products A or B
</ul></p>

<h3>Kinetic model</h3>
<p>The steady state rates may be calculated from: </p>
 <p>$\frac{\partial c_{\text{ESa}}(t)}{\partial t}=k_{32} c_{\text{Enz}}(t) c_{\text{Pa}}(t)+k_{12} c_{\text{Enz}}(t) c_{\text{Sa}}(t)-k_{21} c_{\text{ESa}}(t)-k_{23} c_{\text{ESa}}(t)$</p>
 <p>$\frac{\partial c_{\text{ESb}}(t)}{\partial t}=k_{54} c_{\text{Enz}}(t) c_{\text{Pb}}(t)+k_{14} c_{\text{Enz}}(t) c_{\text{Sb}}(t)-k_{41} c_{\text{ESb}}(t)-k_{45} c_{\text{ESb}}(t)$</p>
 <p>$\frac{\partial c_{\text{Enz}}(t)}{\partial t}=-k_{12} c_{\text{Enz}}(t) c_{\text{Sa}}(t)-k_{14} c_{\text{Enz}}(t) c_{\text{Sb}}(t)+k_{21} c_{\text{ESa}}(t)+k_{23} c_{\text{ESa}}(t)+k_{41} c_{\text{ESb}}(t)+k_{45} c_{\text{ESb}}(t)$</p>

</p>

<p>For steady state conditions ($\frac{\partial c_{\text{i}}(t)}{\partial t} = 0$), the computer solves the equation for $c_{Sa}(t)$, $c_{Sb}(t)$, $c_{enz}(t)$, and $Enz_{total} = c_{E}(t) + c_{ESa}(t) + c_{ESb} $:</p>
    <p>$c_{ESa} = \frac{\left(k_{41}+k_{45}+1\right) \text{Enz}_{\text{total}} \left(-k_{32} c_{\text{Pa}}(t)-k_{12} c_{\text{Sa}}(t)\right)}{\left(k_{41}+k_{45}+1\right) \left(-k_{32} c_{\text{Pa}}(t)-k_{12} c_{\text{Sa}}(t)\right)-\left(k_{21}+k_{23}+1\right) \left(k_{54} c_{\text{Pb}}(t)+k_{14} c_{\text{Sb}}(t)+k_{41}+k_{45}+1\right)}$</p>
    <p>$c_{ESb} = \frac{\text{Enz}_{\text{total}} \left(k_{21} k_{54} c_{\text{Pb}}(t)+k_{23} k_{54} c_{\text{Pb}}(t)+k_{54} c_{\text{Pb}}(t)+k_{14} c_{\text{Sb}}(t)+k_{14} k_{21} c_{\text{Sb}}(t)+k_{14} k_{23} c_{\text{Sb}}(t)\right)}{k_{32} c_{\text{Pa}}(t)+k_{32} k_{41} c_{\text{Pa}}(t)+k_{32} k_{45} c_{\text{Pa}}(t)+k_{54} k_{21} c_{\text{Pb}}(t)+k_{23} k_{54} c_{\text{Pb}}(t)+k_{54} c_{\text{Pb}}(t)+k_{12} c_{\text{Sa}}(t)+k_{12} k_{41} c_{\text{Sa}}(t)+k_{12} k_{45} c_{\text{Sa}}(t)+k_{14} k_{21} c_{\text{Sb}}(t)+k_{14} c_{\text{Sb}}(t)+k_{14} k_{23} c_{\text{Sb}}(t)+k_{41} k_{21}+k_{45} k_{21}+k_{21}+k_{23}+k_{23} k_{41}+k_{41}+k_{23} k_{45}+k_{45}+1}$</p>
    <p>$c_{Enz} = \frac{\left(k_{21}+k_{23}+1\right) \left(k_{41}+k_{45}+1\right) \text{Enz}_{\text{total}}}{k_{32} c_{\text{Pa}}(t)+k_{32} k_{41} c_{\text{Pa}}(t)+k_{32} k_{45} c_{\text{Pa}}(t)+k_{54} k_{21} c_{\text{Pb}}(t)+k_{23} k_{54} c_{\text{Pb}}(t)+k_{54} c_{\text{Pb}}(t)+k_{12} c_{\text{Sa}}(t)+k_{12} k_{41} c_{\text{Sa}}(t)+k_{12} k_{45} c_{\text{Sa}}(t)+k_{14} k_{21} c_{\text{Sb}}(t)+k_{14} c_{\text{Sb}}(t)+k_{14} k_{23} c_{\text{Sb}}(t)+k_{41} k_{21}+k_{45} k_{21}+k_{21}+k_{23}+k_{23} k_{41}+k_{41}+k_{23} k_{45}+k_{45}+1}$</p>
</p>

<p>These expressions involve dependencies on substrate concentrations that we would like to neglect. This is justified because in practice, initial $k_{obs}$ values are used that are measured at still very low product concentrations. Alternatively, one may consider the product release as irreversible ($k_{23}$ and $k_{54}=0$). This gives the following expressions:</p>
<p>$c_{\text{ESa}}(t)= \text{Enz}_{\text{total}}-\frac{\left(k_{21}+k_{23}+1\right) \left(k_{41}+k_{45}+1\right) \text{Enz}_{\text{total}}}{\left(k_{41}+k_{45}+1\right) \left(k_{12} c_{\text{Sa}}(t)+k_{21}+k_{23}+1\right)+k_{14} \left(k_{21}+k_{23}+1\right) c_{\text{Sb}}(t)}-\frac{k_{14} \left(k_{21}+k_{23}+1\right) \text{Enz}_{\text{total}} c_{\text{Sb}}(t)}{\left(k_{41}+k_{45}+1\right) \left(k_{12} c_{\text{Sa}}(t)+k_{21}+k_{23}+1\right)+k_{14} \left(k_{21}+k_{23}+1\right) c_{\text{Sb}}(t)}$, </p> 
<p>$c_{\text{ESb}}(t)=\frac{k_{14} \left(k_{21}+k_{23}+1\right) \text{Enz}_{\text{total}} c_{\text{Sb}}(t)}{\left(k_{41}+k_{45}+1\right) \left(k_{12} c_{\text{Sa}}(t)+k_{21}+k_{23}+1\right)+k_{14} \left(k_{21}+k_{23}+1\right) c_{\text{Sb}}(t)}$, and </p>
<p>$c_{\text{Enz}}(t)=\frac{\left(k_{21}+k_{23}+1\right) \left(k_{41}+k_{45}+1\right) \text{Enz}_{\text{total}}}{\left(k_{41}+k_{45}+1\right) \left(k_{12} c_{\text{Sa}}(t)+k_{21}+k_{23}+1\right)+k_{14} \left(k_{21}+k_{23}+1\right) c_{\text{Sb}}(t)}$</p>

<p>Next, we have to define how we would like to define $k_{obs}$. We choose to sum up the production of product A and B, which may be calculated as $k_{23} \cdot c_{ESa}(t) + k_{45} \cdot c_{ESb}$. Per enzyme we find:</p>
<p>$k_{\text{obs}}=\frac{\frac{k_{45} \left(k_{14} \left(k_{21}+k_{23}+1\right) \text{Enz}_{\text{total}} c_{\text{Sb}}(t)\right)}{\left(k_{41}+k_{45}+1\right) \left(k_{12} c_{\text{Sa}}(t)+k_{21}+k_{23}+1\right)+k_{14} \left(k_{21}+k_{23}+1\right) c_{\text{Sb}}(t)}+k_{23} \left(-\frac{\left(k_{21}+k_{23}+1\right) \left(k_{41}+k_{45}+1\right) \text{Enz}_{\text{total}}}{\left(k_{41}+k_{45}+1\right) \left(k_{12} c_{\text{Sa}}(t)+k_{21}+k_{23}+1\right)+k_{14} \left(k_{21}+k_{23}+1\right) c_{\text{Sb}}(t)}-\frac{k_{14} \left(k_{21}+k_{23}+1\right) \text{Enz}_{\text{total}} c_{\text{Sb}}(t)}{\left(k_{41}+k_{45}+1\right) \left(k_{12} c_{\text{Sa}}(t)+k_{21}+k_{23}+1\right)+k_{14} \left(k_{21}+k_{23}+1\right) c_{\text{Sb}}(t)}+\text{Enz}_{\text{total}}\right)}{\text{Enz}_{\text{total}}}$</p>
<p>$= \frac{\left(k_{41}+k_{45}+1\right) k_{12} k_{23} c_{\text{Sa}}(t)+\left(k_{21}+k_{23}+1\right) k_{14} k_{45} c_{\text{Sb}}(t)}{\left(k_{41}+k_{45}+1\right) \left(k_{12} c_{\text{Sa}}(t)+k_{21}+k_{23}+1\right)+\left(k_{21}+k_{23}+1\right) k_{14} c_{\text{Sb}}(t)}$.</p>
<p>There is a symmetry in this formula that becomes even more obvious if we substitute $\text{iA}=k_{21}+k_{23}+1$ and $\text{iB}=k_{41}+k_{45}+1$:</p>
<p>$\Longrightarrow k_{\text{obs}}=\frac{\text{iB} k_{12} k_{23} c_A+\text{iA} k_{14} k_{45} c_B}{\frac{k_{12} c_A}{\text{iA}}+\frac{k_{14} c_B}{\text{iB}}+1}$. The symmetry for $c_{Sa}$ and $c_{Sb}$ is visible in the following plot.</p>

```python
%matplotlib widget

import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider, SelectionSlider
```

```python
import numpy as np
import matplotlib.pyplot as plt

from ipywidgets import (
    interact,
    FloatSlider,
    SelectionSlider
)


def k_obs(
    x1,
    x2,
    k12=5.0,
    k21=0.5,
    k23=25.0,
    k32=0.0,
    k14=5.0,
    k41=0.5,
    k45=1.0,
    k54=0.0,
):
    """
    Observed rate constant for competing-substrate kinetics.

    Parameters
    ----------
    x1 : array-like
        Concentration of substrate A.
    x2 : array-like
        Concentration of substrate B.

    Returns
    -------
    array-like
        Observed rate constant k_obs in s^-1.

    Notes
    -----
    k32 is retained as a parameter to match the Mathematica interface,
    but it does not enter the current expression because all terms
    containing k32 were commented out in the Mathematica code.
    """

    numerator = (
        k12 * k45 * (k54 * x2 + k14 * x2)
        + k23
        * (
            k12 * k41 * x1
            + k45 * (k12 * x1 + k14 * x2)
        )
    )

    denominator = (
        (k41 + k45) * k12 * x1
        + k21 * (k41 + k45 + k14 * x2)
        + k23 * (k41 + k45 + k14 * x2)
    )

    # Avoid division warnings at singular parameter combinations.
    return np.divide(
        numerator,
        denominator,
        out=np.full_like(np.asarray(numerator, dtype=float), np.nan),
        where=np.abs(denominator) > 1e-14,
    )


def plot_competing_substrates(
    k12=5.0,
    k21=0.5,
    k23=25.0,
    k32=0.0,
    k14=5.0,
    k41=0.5,
    k45=1.0,
    k54=0.0,
    inset_x=0.10,
    inset_y=0.10,
    font_size=18,
    image_size=720,
):
    plt.close("all")

    # Concentration grid
    x1 = np.linspace(0.0, 10.0, 121)
    x2 = np.linspace(0.0, 10.0, 121)
    X1, X2 = np.meshgrid(x1, x2)

    Z = k_obs(
        X1,
        X2,
        k12=k12,
        k21=k21,
        k23=k23,
        k32=k32,
        k14=k14,
        k41=k41,
        k45=k45,
        k54=k54,
    )

    # Mathematica PlotRange upper limit
    second_limit_denominator = k14 * (k21 + k23)

    if abs(second_limit_denominator) > 1e-14:
        second_limit = (
            k45
            * (k14 * k23 + k12 * (k14 + k54))
            / second_limit_denominator
        )
    else:
        second_limit = 0.0

    z_max = max(k23, second_limit)

    # Convert nominal pixel size to Matplotlib inches
    dpi = 100
    figure_width = image_size / dpi
    figure_height = 0.78 * figure_width

    fig = plt.figure(
        figsize=(figure_width, figure_height),
        dpi=dpi,
        constrained_layout=True,
    )

    ax = fig.add_subplot(111, projection="3d")

    ax.plot_surface(
        X1,
        X2,
        Z,
        rstride=1,
        cstride=1,
        cmap="viridis",
        linewidth=0,
        antialiased=True,
    )

    ax.set_xlim(0.0, 10.0)
    ax.set_ylim(0.0, 10.0)
    ax.set_zlim(0.0, z_max)

    ax.set_xlabel(
        r"$c_{\mathrm{Substrate\ A}}$",
        fontsize=font_size,
        fontweight="bold",
        labelpad=12,
    )
    ax.set_ylabel(
        r"$c_{\mathrm{Substrate\ B}}$",
        fontsize=font_size,
        fontweight="bold",
        labelpad=12,
    )
    ax.set_zlabel(
        r"$k_{\mathrm{obs}}\;[\mathrm{s}^{-1}]$",
        fontsize=font_size,
        fontweight="bold",
        labelpad=12,
    )

    ax.tick_params(
        axis="both",
        which="major",
        labelsize=max(font_size - 5, 8),
    )

    ax.view_init(elev=28, azim=-135)

    parameter_text = (
        rf"$K_{{Sa}} = k_{{21}}/k_{{12}} = {k21 / k12:.4g}$"
        "\n"
        rf"$k_{{23}} = {k23:.4g}$"
        "\n"
        rf"$K_{{Sb}} = k_{{41}}/k_{{14}} = {k41 / k14:.4g}$"
        "\n"
        rf"$k_{{45}} = {k45:.4g}$"
    )

    # ax.text2D uses axes-relative coordinates.
    ax.text2D(
        inset_x,
        inset_y,
        parameter_text,
        transform=ax.transAxes,
        fontsize=max(font_size - 4, 7),
        horizontalalignment="left",
        verticalalignment="bottom",
        bbox={
            "boxstyle": "round,pad=0.4",
            "facecolor": "white",
            "alpha": 0.80,
            "edgecolor": "0.6",
        },
    )

    plt.show()


interactive_plot = interact(
    plot_competing_substrates,

    k12=FloatSlider(
        value=5.0,
        min=0.01,
        max=100.0,
        step=0.01,
        description="k₁₂",
        continuous_update=False,
    ),

    k21=FloatSlider(
        value=0.5,
        min=0.01,
        max=100.0,
        step=0.01,
        description="k₂₁",
        continuous_update=False,
    ),

    k23=FloatSlider(
        value=25.0,
        min=0.0,
        max=100.0,
        step=0.1,
        description="k₂₃",
        continuous_update=False,
    ),

    k32=FloatSlider(
        value=0.0,
        min=0.0,
        max=100.0,
        step=0.1,
        description="k₃₂",
        continuous_update=False,
    ),

    k14=FloatSlider(
        value=5.0,
        min=0.01,
        max=100.0,
        step=0.01,
        description="k₁₄",
        continuous_update=False,
    ),

    k41=FloatSlider(
        value=0.5,
        min=0.01,
        max=100.0,
        step=0.01,
        description="k₄₁",
        continuous_update=False,
    ),

    k45=FloatSlider(
        value=25.0,
        min=0.0,
        max=100.0,
        step=0.1,
        description="k₄₅",
        continuous_update=False,
    ),

    k54=FloatSlider(
        value=0.0,
        min=0.0,
        max=100.0,
        step=0.1,
        description="k₅₄",
        continuous_update=False,
    ),

    inset_x=FloatSlider(
        value=0.10,
        min=0.0,
        max=1.0,
        step=0.01,
        description="Inset x",
        continuous_update=False,
    ),

    inset_y=FloatSlider(
        value=0.10,
        min=-1.0,
        max=1.0,
        step=0.01,
        description="Inset y",
        continuous_update=False,
    ),

    font_size=SelectionSlider(
        options=[9, 12, 14, 18, 24],
        value=18,
        description="Font size",
        continuous_update=False,
    ),

    image_size=SelectionSlider(
        options=[450, 650, 720, 950, 1080, 1280],
        value=720,
        description="Image size",
        continuous_update=False,
    ),
)
```

<h3>Plot legend</h3>
<p>The x-axis takes the concentration of substrate A, $c_{S,A}$ as free variable, the y-axis that of substrate B, $c_{S,B}$. The z-axis calculates the observed rate $k_{obs}$. Each of the substrate dependencies ($k_{obs}$ vs $c_A$, or $k_{obs}$ vs $c_B$) gives a typical, hyperbolic curve. Consider the following limiting cases:</p>
<p>(a)Assuming that substrate B is a competitive inhibitor, i.e. $k_{45} \ll k_{23} $ or even zero, each given concentration of substrate A follows an inverse hyperbolic dependence on increasing inhibitor concentrations $c_B$.</p>
<p>(b) Assuming $S_B$ is a fluorescent analog of $S_A$, it might be hydrolyzed equally efficiently as $S_A$ but show a lower affinity ($\frac{k_{21}}{k_{12}} < \frac{k_{41}}{k_{14}}$). The Michaelis-Menten hyperbolas on the x-axis might thus show a lower half-maximal activation constant $K_{1/2}$.</p>
